```python
from turtle import Turtle, Screen

timmy = Turtle()

# we lean about color to change and shape of turtle
timmy.shape("turtle")
timmy.color("red","green")

# How to move a turtle
timmy.forward(100)
timmy.right(91)
timmy.forward(100)
timmy.right(91)
timmy.forward(100)
timmy.right(50)
timmy.forward(100)




screen = Screen()
screen.exitonclick()
```

```python
import turtle

t = turtle.Turtle()

# Draw Square
for i in range(4):
    t.forward(100)
    t.right(90)

# Move to new position
t.penup()
t.goto(150, 0)
t.pendown()

# Draw Circle
t.circle(50)

turtle.done()
```

```python
import turtle

t = turtle.Turtle()

# Draw rectangle
t.forward(200)   # length
t.right(90)

t.forward(100)   # width
t.right(90)

t.forward(200)
t.right(90)

t.forward(100)

turtle.done()
```

```python
import turtle

t = turtle.Turtle()

for i in range(15):   # number of dashes
    t.forward(10)     # draw line
    t.penup()         # lift pen (gap)
    t.forward(10)     # move forward (space)
    t.pendown()       # put pen down

turtle.done()
```

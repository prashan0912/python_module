```python
# Use the global keyword inside a function when you want to modify a variable that was defined outside the function (at the global scope).

count = 0

def increase():
    global count
    count += 1   # ✅ modifies global variable

increase()
print(count)  # 1
```

```python
def is_prime(num):
    if num <= 1:
        return False
    
    for i in range(2, num):
        if num % i == 0:
            return False
    
    return True
    
```

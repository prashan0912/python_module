```python
 # positional argument

def function1(location , name):
    print("location : ",location , "name : ", name) 


name = "Rahul";
location = "Delhi"
print(name , location)
function1(name , location)
```

```python
 # positional argument

def function1(p,q):
    print(q-p+7) 


a = 10;
b = 20
print(a,b)
function1(a,b)
```

```python
 # positional argument

def function1(location , name):
    print("location : ",location , "name : ", name) 


name = "Rahul";
location = "Delhi"
print(name , location)
function1(name = name , location =location)
```

```python
 # keyword argument

def function1(a,b):
    print(b-a+7) 

a = 10;
b = 20
print(a,b)
function1(b=10,a=20)
```

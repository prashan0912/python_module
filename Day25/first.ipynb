```python
import pandas as pd

# Sample DataFrame
df = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['NY', 'LA', 'Chicago']
})

# Convert to dictionary (default)
dict_default = df.to_dict()
print("Default orient:\n", dict_default)

# Convert to list format
dict_list = df.to_dict(orient='list')
print("\nList orient:\n", dict_list)

# Convert to records format
dict_records = df.to_dict(orient='records')
print("\nRecords orient:\n", dict_records)
```
